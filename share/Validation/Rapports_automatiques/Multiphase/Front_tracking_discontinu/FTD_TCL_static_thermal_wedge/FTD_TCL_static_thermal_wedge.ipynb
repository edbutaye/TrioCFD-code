### Description
 
Computation of heat transfer in the vicinity of a contact-line

Data analyzed
* integrated wall heat flux
* integrated interfacial flux

Assumptions:
* No effect of phase-change on the interface (no motion)
* Prescribed wall temperature
 


### TRUST parameters
 
* Version TRUST : 1.9.4 
* Type of problem : Probleme_FT_Disc_gen 
* Discretization : VDF 
* Time scheme : Euler_Explicite 
* Solving of equations : Navier_stokes_FT_disc without turbulence Model 
* Solving of equations : Convection_Diffusion_Temperature_FT_Dist for liquid 
* Convection scheme : Momentum \& Temperature : quick 
* Diffusion scheme : Default {} for all equations 
* No Turbulence model : DNS 
* Type of boundary conditions : Heated wall with constant temperature and prescribed contact angle (bottom), free openings or walls otherwise. 
* Type of fluid : Atm. water\&steam.

## Sensitivity to the offset
 
How does the flux evolves when the contact line crosses to a new Eulerian cell at the wall? 
Strangely, all calculations are stables except OFFSET-0.2 which requires to set the TIMESTEP to 2e-07 and OFFSET0.2 which is even more demanding (dt=1e-7). 
 Remark: The integral flux  (shown in Chap 4.2) converges consistently regardless of the offset, demonstrating the model's conservation behavior, which is a important feature for the moving triple contact line.
 In the reference case, the leftward shift of the front (TCL) occurs due to remeshing, which we cannot eliminate because it would lead to the marker being situated at the grid face, causing issues with the employed algorithm (*.med) in post-processing. This shift slightly alters the simulated contact angle. Additionally, due to this leftward movement, only a smaller portion of the front enters the adjacent left cell. Consequently, a reduced part of Qmicro is assigned to this cell, which ideally should be entirely allocated to the second cell. Despite this effect, we believe its overall influence on the system's behavior remains limited. This assumption is supported by the convergence of the integral flux, indicating that despite the leftward shift and the altered distribution of Qmicro, the system's integral behavior remains consistent and converges as expected.
 


```python
from trustutils import plot 

def myplot(fig, icol,ylab, subtit, log=False, xmm=(0., 20.)):
    data = plot.loadText("CAS_OFFSET-0.5/xphi_last.txt")
    fig.count+=1
    ix = int(fig.count<2)
    iy = fig.count%2
    fig.addPlot((ix,iy),subtit)
    for i, offset in enumerate(cases):
        data = ldata[i]
        fig.add((1e6*(data[0]-5e-6-float(offset)*1e-6)),(data[icol]),label=rf"offset: {offset}",marker='-x')

    fig.label(r"d [$\mu$m] (dist. from TCL)",ylab)
    fig.visu(xmin=xmm[0],xmax=xmm[1])
    
    if log: fig.scale(yscale='log')
    return fig

cases = [f"{x:.1f}" for x in np.arange(-0.5, 0.6,0.1)]
ldata = []
for i, offset in enumerate(cases):
    o = offset
    if (offset == "-0.2"): o+="_TIMESTEP2e-07"
    if (offset == "0.2"): o+="_TIMESTEP1e-07"
    if (offset == "-0.0"): o="0.0"
    data = plot.loadText(f"CAS_OFFSET{o}/xphi_last.txt")
    ldata.append(data)

dic = {}
dic["phi"] = (1,r"$\phi$ [W/m²]")
dic["pui"] = (2,r"$P$ [W]")
for v in dic.keys():
    icol, lab = dic[v]
    if (v=="phi"):
        fig = plot.Graph(title=r"(final) flux at the bottom wall",subtitle="global", nX=2,nY=2)
    else:
        fig = plot.Graph(title=r"(final) flux at the bottom wall",subtitle="global", nX=2,nY=2)
    fig.count = -1
    for log in [False, True]:
        st = ""
        for j, xmm in enumerate([(0., 20.), (0., 3.)]):
            if (j==1): st="Zoom"
            fig = myplot(fig, icol,ylab=lab, subtit=st, log=log, xmm=(0., 20.))
```

## Effect of the microscopic contact angle
 
The microscopic contact angle is varied from 5 to 90 degrees.
Firstly, In the generated figure for angle 5, only the results for the micro and meso zones are depicted, considering the utilization of 4 meso cells. For the flux show in the figure, larger contact angles display a curve representing micro, meso, and macro zones, whereas smaller contact angles only exhibit micro and meso zones.
 The flux demonstrates a distinct pattern for smaller contact angles: it initially decreases in the micro-zone, then exhibits an increase during the transition from the micro to the meso zone, followed by a consistent decrease within the meso zone.  This peculiar transition behavior is a result of the interfacial heat resistance employed in the micro-zone. For a smooth and coherent flux transition (where the flux consistently decreases moving away from the triple contact line), it's crucial to ensure that the flux in the micro-zone ($q\_\{micro\}$) is less than the flux in the meso zone ($q\_\{meso\}$ at sm+$\Delta x$). This condition necessitates $\Delta x $ bigger than $R\_i \lambda\_\{l\} / \theta\_\{app\}$ to maintain a proper transient flux.
 In most cases, excluding angles 5, 80, and 90, the front shifts leftward due to remeshing, slightly altering the simulated contact angle. When only a smaller section of the front enters the adjacent left cell, the flux distribution is affected (as seen in angle 50).
 


```python
from trustutils import plot 
def doStuff(i):
    ls = ['-+', '-.v', '-o', '-', '-x', '--o', ':', ':x']
    for j, ang in enumerate([5, 10, 15, 20, 30, 50, 80, 90]):
        data = plot.loadText(f"CAS_ANGLE{ang}/xphi_last.txt")
        v = str(ang)
        if (v == "50"): 
            v+=" (ref)"
        fig.add((1e6*(data[0]-5e-6+0.e-6)),(data[i+1])*ang,label=r"angle: %s"%v,marker=ls[j])
    if i==0:
        fig.label(r"d [$\mu$m] (dist. from TCL)",r"$\theta_{app}\times \phi$ $[W/m².^\circ]$")
    else:
        fig.label(r"d [$\mu$m] (dist. from TCL)",r"$\theta_{app}\int_0^s \phi ds$ [W/m.^\circ]")

for i, name in enumerate(["local", "integrated"]):
    fig = plot.Graph(f"(final) {name} flux along the bottom wall", nX=2, nY=2) 
        
    fig.addPlot((0,0),"std")
    doStuff(i)
    fig.visu(xmin=0,xmax=20.)

    fig.addPlot((0,1),"zoom")
    doStuff(i)
    fig.visu(xmin=0,xmax=3.)

    fig.addPlot((1,0),"log")
    doStuff(i)
    fig.scale(yscale='log')
    fig.visu(xmin=0,xmax=20.)

    fig.addPlot((1,1),"log zoom")
    doStuff(i)
    fig.scale(yscale='log')
    fig.visu(xmin=0,xmax=3.)
```

## Effect of the wall overheat
 
The wall overheat is varied from 3 to 15$\deg C$.
In all cases, the leftward shift of the front (TCL) occurs due to remeshing, which we cannot eliminate because it would lead to the marker being situated at the grid face, causing issues with the employed algorithm (*.med) in post-processing. This shift slightly alters the simulated contact angle. Additionally, due to this leftward movement, only a smaller portion of the front enters the adjacent left cell. Consequently, a reduced part of Qmicro is assigned to this cell, which ideally should be entirely allocated to the second cell. Despite this effect, we believe its overall influence on the system's behavior remains limited. This assumption is supported by the convergence of the integral flux, indicating that despite the leftward shift and the altered distribution of Qmicro, the system's integral behavior remains consistent and converges as expected. (see Chap. 4.2)
