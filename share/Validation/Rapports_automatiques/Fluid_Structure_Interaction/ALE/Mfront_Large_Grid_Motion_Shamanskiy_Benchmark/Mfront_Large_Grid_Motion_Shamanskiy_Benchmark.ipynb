### Initial Conditions and Boundary Conditions 
Boundary conditions for the fluid problem are no-slip wall conditions on the top and bottom horizontal borders and zero pressure conditions on the vertical and oblique borders. The borders of the tab inside the domain are Lagrangian boundaries with an oscillatory imposed motion along the first bending mode given by:

\begin{equation}
    \begin{aligned}
        {\mathbf{u}_{imp}}_x &= -a \times  \frac{4y}{3l^4}(x^2-3lx+3l^2) \times sin(f \times 2\pi t), \\
        {\mathbf{u}_{imp}}_y &= a \times \frac{x^2}{3l^4}(x^2-4lx+6l^2) \times sin(f \times 2\pi t),
    \end{aligned}
\end{equation}
where $f$ is the frequency arbitrarily set to $10$ Hz and $a$ is a scaling parameter set to $0.20$ mm controlling the maximum deflection of the tip of the tab.


### Fluid Properties 

The fluid properties are $\rho = 1$ kg.m$^{-3}$ and $\nu = 0.02$ m$^2$.s$^{-1}$.

### Case Setup

 The Ogden model is chosen for the grid motion problem, with the first set of parameters $\alpha = 28.8$, $\mu = 5000$ Pa, and $K = 3576000$ Pa. The corresponding Poisson ratio, relevant to measure the level of incompressibility of the material model, is deduced from the formulae:

\begin{equation}
    %\begin{aligned}
         \tilde{\mu} = \frac{1}{2} \left ( \frac{\mu}{2} \times \frac{\alpha}{2} \right) \text{ (initial shear modulus)}, \quad \text{and}\quad
         K = \tilde{\mu} \frac{2 (1 + \nu)}{3 (1-2\nu)}.
    %\end{aligned}
\end{equation}

This yields a value of the Poisson ratio $\mu = 0.497$ in the present case, showing the capability of the proposed approach to handle materials very close to incompressibility. The initial fictitious grid density is set to $10^4$ kg.m$^{-3}$ and mass scaling is activated to prevent the grid stability time step from going below $10^{-7}$ s, keeping the time step ratio between the grid problem and the physical problem below $200$ even in the extremum position of the tab for the sake of computational efficiency.

## Conclusion 
 
Using the Ogden material model yields a very robust mesh motion for this particular case since all sets of parameters manage to handle significant distortions and preserve an aspect ratio for each cell compatible with the solution of the physical fluid problem.
Results at time $0.025$ s match those in the reference paper [1] in terms of preservation of the cells aspect ratio throughout the mesh deformation process.
